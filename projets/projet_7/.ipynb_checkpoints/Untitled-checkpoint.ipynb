## data

```python
csv_path = 'data'
file_names = [x for x in os.listdir(csv_path) if ".csv" in x]
file_names
```

```python
# Liste des fichiers CSV à charger
files = [
    'data/application_test.csv',
    'data/HomeCredit_columns_description.csv',
    'data/application_train.csv',
    'data/installments_payments.csv',
    'data/bureau_balance.csv',
    'data/POS_CASH_balance.csv',
    'data/bureau.csv',
    'data/credit_card_balance.csv',
    'data/previous_application.csv',
    'data/sample_submission.csv'
]

# Dictionnaire pour stocker les DataFrames
dataframes = {}

# Charger chaque fichier CSV et vérifier s'il est correctement chargé
for file in files:
    try:
        df = pd.read_csv(file, encoding='ISO-8859-1')
        if df is not None:
            dataframes[file] = df
            print(f"{file} chargé avec succès.")
        else:
            print(f"Erreur lors du chargement de {file}.")
    except Exception as e:
        print(f"Erreur lors du chargement de {file}: {e}")

# Supprimer la colonne "Unnamed: 0" si elle existe dans homecredit_columns_description
if '/data/HomeCredit_columns_description.csv' in dataframes:
    if 'Unnamed: 0' in dataframes['/data/HomeCredit_columns_description.csv'].columns:
        dataframes['/data/HomeCredit_columns_description.csv'].drop(columns=['Unnamed: 0'], inplace=True)

# Afficher les premières lignes de chaque DataFrame pour vérifier
for file, df in dataframes.items():
    print(f"Premières lignes de {file}:")
    print(df.head())
```

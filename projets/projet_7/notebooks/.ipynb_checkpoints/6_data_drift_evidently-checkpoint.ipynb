# Data Drift

```python
import pandas as pd
from evidently import Report
from evidently.presets import DataDriftPreset
```

```python
csv_path_output = '../data/output/'
#X_train_filename = 'train_df_cleaned_smote.parquet.gzip'
X_train_filename = 'train_df_cleaned.parquet.gzip'
X_test_filename = 'test_df_cleaned.parquet.gzip'

X_train = pd.read_parquet(csv_path_output + X_train_filename)
X_test_kag = pd.read_parquet(csv_path_output + X_test_filename)
```

```python
X_train.shape
```

```python
X_train = X_train.drop(columns=['SK_ID_CURR','TARGET'])
X_test_kag = X_test_kag.drop(columns=['EXT_SOURCE_1', 'SK_ID_CURR'])

print(X_train.shape)
print(X_test_kag.shape)
```

```python
report = Report([
    DataDriftPreset(method="psi")
],
include_tests="True")
my_eval = report.run(X_train, X_test_kag)
my_eval
```

```python
# Fonction pour appeler l'API externe
def predict_target(data):
    # Remplacez l'URL par celle de votre API
    api_url = "http://votre-api.com/predict"
    response = requests.post(api_url, json=data)
    return response.json()

# Charger les données
@st.cache
def load_data():
    csv_path_output = '../data/output/'
    X_train_filename = 'train_df_cleaned_smote.parquet.gzip'
    X_test_filename = 'test_df_cleaned.parquet.gzip'
    X_train = pd.read_parquet(csv_path_output + X_train_filename)
    X_test = pd.read_parquet(csv_path_output + X_test_filename)
    return X_train, X_test

X_train, X_test = load_data()

# Interface Streamlit
st.title("Dashboard de Prédiction de Crédit")

# Afficher les données d'entraînement
st.header("Données d'entraînement")
st.write(X_train)

# Sélectionner un individu à évaluer
st.header("Évaluer un nouvel individu")
individual_index = st.selectbox("Sélectionnez un individu", X_test.index)

# Afficher les informations de l'individu sélectionné
st.write(X_test.loc[individual_index])

# Bouton pour lancer la prédiction
if st.button("Lancer la prédiction"):
    data = X_test.loc[individual_index].to_dict()
    prediction = predict_target(data)
    st.write(f"Prédiction: {prediction}")

    # Afficher les individus proches avec TARGET à 1
    st.header("Individus proches avec TARGET à 1")
    # Ajoutez ici la logique pour trouver et afficher les individus proches
```

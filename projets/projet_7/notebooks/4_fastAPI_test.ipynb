# Tests points d'entrée

## /data

```python
# Créer le répertoire s'il n'existe pas
output_dir = "../data/api_output"
os.makedirs(output_dir, exist_ok=True)
```

```python
# Format parquet
response = requests.get("http://127.0.0.1:8000/data?format=parquet")

# Afficher les 100 premiers octets pour un aperçu
preview = response.content[:100]
print("Aperçu des données Parquet :", preview)

# Écrire le contenu complet dans un fichier dans le répertoire spécifique
parquet_output_path = os.path.join(output_dir, "data.parquet")
with open(parquet_output_path, "wb") as f:
    f.write(response.content)
```

```python
# Format CSV
response = requests.get("http://127.0.0.1:8000/data?format=csv")

# Afficher les 100 premiers octets pour un aperçu
preview = response.content[:100]
print("Aperçu des données CSV :", preview.decode('utf-8'))  # Décoder en chaîne de caractères pour CSV

# Écrire le contenu complet dans un fichier dans le répertoire spécifique
csv_output_path = os.path.join(output_dir, "data.csv")
with open(csv_output_path, "wb") as f:
    f.write(response.content)
```

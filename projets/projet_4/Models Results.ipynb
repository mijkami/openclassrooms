```python
import pandas as pd
```

```python
csv_name_1 = 'energy_no_scarscore_performances.csv'
csv_name_2 = 'energy_with_scarscore_performances.csv'
csv_name_3 = 'emissions_no_scarscore_performances.csv'
csv_name_4 = 'emissions_with_scarscore_performances.csv'
```

```python
df1 = pd.read_csv(csv_name_1, low_memory=False)
df2 = pd.read_csv(csv_name_2, low_memory=False)
df3 = pd.read_csv(csv_name_3, low_memory=False)
df4 = pd.read_csv(csv_name_4, low_memory=False)
```

```python
df_results = pd.concat([df1, df2, df3, df4], axis=0, ignore_index=True)
```

```python
df_results.to_csv('models_results_no_WB.csv', index=False)
```

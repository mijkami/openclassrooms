# Imports

```python
import pandas as pd

import requests
import json
#import sgqlc

import data.credentials as creds
```

# Query API

```python
# https://rapidapi.com/edamam/api/edamam-food-and-grocery-database/playground/apiendpoint_332c01e7-e404-4439-bf69-0b00c00ef348

url = "https://edamam-food-and-grocery-database.p.rapidapi.com/api/food-database/v2/parser"

querystring = {
    "ingr":"champagne"
}

headers = {
	"x-rapidapi-key": creds.api_key,
	"x-rapidapi-host": "edamam-food-and-grocery-database.p.rapidapi.com"
}
```

```python
response = requests.get(url, headers=headers, params=querystring)
json_data = json.loads(response.text)
json_data
```

# Extract Transform Data

```python
# Extracting the relevant information from the JSON data
data = {
    "foodId": [],
    "label": [],
    "category": [],
    "foodContentsLabel": [],
    "image": []
    #,"brand": []
}

max_items = 10

# Parsing the JSON data
for hint in json_data['hints'][:max_items]:
    food = hint['food']
    data["foodId"].append(food.get("foodId"))
    data["label"].append(food.get("label"))
    data["category"].append(food.get("category"))
    data["foodContentsLabel"].append(food.get("foodContentsLabel", ""))
    data["image"].append(food.get("image", ""))
    #data["brand"].append(food.get("brand", ""))

# Creating a DataFrame
df = pd.DataFrame(data)
df
```

# Export

```python
#export csv
df.to_csv('data/champagne_prods.csv', index=False)
print('Export Done!')
```

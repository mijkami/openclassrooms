### Retrieve categories from 'product_category_tree 

```python
def get_categories_from_series(series):
    
    # Function to get the 3 first cat
    def get_categories_from_cell(cell):
        product_category_tree = cell.strip('[]"')
        product_categories_list = product_category_tree.split(" >> ")
        
        categories = pd.Series()
        
        for i in range(min(3, len(product_categories_list))):
            categories["cat_{}".format(i+1)] = product_categories_list[i]
        
        return categories

    categories_df = series.apply(get_categories_from_cell).fillna("")
    return categories_df
```

```python
# cleaning

nlp = spacy.load('en_core_web_sm', disable=['ner', 'parser']) # disabling Named Entity Recognition for speed

def cleaning(doc):
    # Lemmatizes and removes stopwords
    # doc needs to be a spacy Doc object
    txt = [token.lemma_ for token in doc if not token.is_stop]
    # Word2Vec uses context words to learn the vector representation of a target word,
    # if a sentence is only one or two words long,
    # the benefit for the training is very small
    if len(txt) > 2:
        return ' '.join(txt)
```

```python
brief_cleaning = (re.sub("[^A-Za-z']+", ' ', str(row)).lower() for row in text_2vec)
brief_cleaning
```

```python
t = time()

txt = [cleaning(doc) for doc in nlp.pipe(brief_cleaning, batch_size=5000, n_process=-1)]

print('Time to clean up everything: {} mins'.format(round((time() - t) / 60, 2)))
```

```python
df_clean = pd.DataFrame({'clean': txt})
df_clean = df_clean.dropna().drop_duplicates()
df_clean.shape
```

```python
# bigrams
sent = [row.split() for row in df_clean['clean']]

phrases = Phrases(sent, min_count=40, progress_per=10000)
#phrases = Phrases(sent, min_count=5, progress_per=100)
#phrases = Phrases(sent, min_count=20, progress_per=5000)

bigram = Phraser(phrases)
sentences = bigram[sent]
sentences
```

```python
# most frequent words

word_freq = defaultdict(int)

for sent in sentences:
    for i in sent:
        word_freq[i] += 1
len(word_freq)
```

```python
cores = multiprocessing.cpu_count()
cores
```

- min_count = int - Ignores all words with total absolute frequency lower than this - (2, 100)
- window = int - The maximum distance between the current and predicted word within a sentence. E.g. window words on the left and window words on the left of our target - (2, 10)
- size = int - Dimensionality of the feature vectors. - (50, 300)
- sample = float - The threshold for configuring which higher-frequency words are randomly downsampled. Highly influencial. - (0, 1e-5)
- alpha = float - The initial learning rate - (0.01, 0.05)
- min_alpha = float - Learning rate will linearly drop to min_alpha as training progresses. To set it: alpha - (min_alpha * epochs) ~ 0.00
- negative = int - If > 0, negative sampling will be used, the int for negative specifies how many "noise words" should be drown. If set to 0, no negative sampling is used. - (5, 20)
- workers = int - Use these many worker threads to train the model (=faster training with multicore machines)


```python
word_vec_params = {
        'sentences' : 'text',
        'vector_size' : 200,
        'window' : 5,
        'min_count' : 1,
        'workers' : cores - 1
}
```

```python
import numpy as np
from sklearn.manifold import TSNE
from sklearn.metrics import adjusted_rand_score, confusion_matrix, ConfusionMatrixDisplay
from sklearn.cluster import KMeans
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt

# Ajouter la colonne 'cat_1' de df à df_clean
df_clean['cat_1'] = df['cat_1']

# Étape 1 : Calculer les embeddings pour chaque document
def document_vector(doc, model):
    # Créez un vecteur pour le document en faisant la moyenne des vecteurs de mots
    doc = [word for word in doc if word in model.wv.index_to_key]
    return np.mean(model.wv[doc], axis=0) if len(doc) > 0 else np.zeros(model.vector_size)

# Calculer les embeddings pour chaque document
X = np.array([document_vector(doc, w2vec) for doc in sentences])

# Assurez-vous que df_clean a une colonne avec les étiquettes réelles
assert 'cat_1' in df_clean.columns, "La colonne 'cat_1' n'existe pas dans df_clean."

# Obtenez les étiquettes réelles
y_true = df_clean['cat_1'].values

# Convertir les étiquettes réelles en entiers
label_encoder = LabelEncoder()
y_true_encoded = label_encoder.fit_transform(y_true)

# Vérifiez que X et y_true ont le même nombre d'échantillons
assert X.shape[0] == y_true_encoded.shape[0], "Le nombre d'échantillons ne correspond pas entre X et y_true."

# Étape 2 : Réduire la dimensionnalité avec t-SNE
tsne = TSNE(n_components=2, random_state=42)
X_tsne = tsne.fit_transform(X)

# Étape 3 : Calculer le score ARI
# Utiliser KMeans pour obtenir des clusters à partir des embeddings
kmeans = KMeans(n_clusters=len(l_cat), random_state=42)
y_pred = kmeans.fit_predict(X)

# Comparer les clusters obtenus avec les catégories réelles
ari_score = adjusted_rand_score(y_true_encoded, y_pred)
print(f'ARI Score: {ari_score}')

# Étape 4 : Calculer et afficher la matrice de confusion
cm = confusion_matrix(y_true_encoded, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=label_encoder.classes_)
disp.plot(cmap=plt.cm.Blues)
plt.title('Confusion Matrix')
plt.show()

# Étape 5 : Visualiser les résultats avec t-SNE
plt.figure(figsize=(10, 8))
scatter = plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=y_pred, cmap='viridis', alpha=0.6)
legend1 = plt.legend(*scatter.legend_elements(), title="Clusters")
plt.gca().add_artist(legend1)
plt.title('t-SNE Visualization of Document Embeddings')
plt.xlabel('t-SNE Component 1')
plt.ylabel('t-SNE Component 2')
plt.show()
```

```python
import numpy as np
from sklearn.manifold import TSNE
from sklearn.metrics import adjusted_rand_score, confusion_matrix, ConfusionMatrixDisplay
from sklearn.cluster import KMeans
from sklearn.preprocessing import LabelEncoder
from scipy.optimize import linear_sum_assignment
import matplotlib.pyplot as plt

# Ajouter la colonne 'cat_1' de df à df_clean
df_clean['cat_1'] = df['cat_1']

# Étape 1 : Calculer les embeddings pour chaque document
def document_vector(doc, model):
    # Créez un vecteur pour le document en faisant la moyenne des vecteurs de mots
    doc = [word for word in doc if word in model.wv.index_to_key]
    return np.mean(model.wv[doc], axis=0) if len(doc) > 0 else np.zeros(model.vector_size)

# Calculer les embeddings pour chaque document
X = np.array([document_vector(doc, w2vec) for doc in sentences])

# Assurez-vous que df_clean a une colonne avec les étiquettes réelles
assert 'cat_1' in df_clean.columns, "La colonne 'cat_1' n'existe pas dans df_clean."

# Obtenez les étiquettes réelles
y_true = df_clean['cat_1'].values

# Convertir les étiquettes réelles en entiers
label_encoder = LabelEncoder()
y_true_encoded = label_encoder.fit_transform(y_true)

# Vérifiez que X et y_true ont le même nombre d'échantillons
assert X.shape[0] == y_true_encoded.shape[0], "Le nombre d'échantillons ne correspond pas entre X et y_true."

# Étape 2 : Réduire la dimensionnalité avec t-SNE
tsne = TSNE(n_components=2, random_state=42)
X_tsne = tsne.fit_transform(X)

# Étape 3 : Calculer le score ARI
# Utiliser KMeans pour obtenir des clusters à partir des embeddings
kmeans = KMeans(n_clusters=len(l_cat), random_state=42)
y_pred = kmeans.fit_predict(X)

# Comparer les clusters obtenus avec les catégories réelles
ari_score = adjusted_rand_score(y_true_encoded, y_pred)
print(f'ARI Score: {ari_score}')

# Étape 4 : Calculer la matrice de confusion initiale
cm = confusion_matrix(y_true_encoded, y_pred)

# Étape 5 : Diagonaliser la matrice de confusion
row_ind, col_ind = linear_sum_assignment(-cm)
cm_diagonalized = cm[:, col_ind]

# Étape 6 : Afficher la matrice de confusion diagonalise avec les étiquettes inclinées
disp = ConfusionMatrixDisplay(confusion_matrix=cm_diagonalized, display_labels=[label_encoder.classes_[i] for i in col_ind])
disp.plot(cmap=plt.cm.Blues, xticks_rotation=55)  # Incliner les étiquettes à 55 degrés
plt.title('Diagonalized Confusion Matrix')
plt.show()

# Étape 7 : Visualiser les résultats avec t-SNE
fig, axes = plt.subplots(1, 2, figsize=(20, 8))

# Sous-graphique 1 : Projection des catégories réelles
scatter1 = axes[0].scatter(X_tsne[:, 0], X_tsne[:, 1], c=y_true_encoded, cmap='viridis', alpha=0.6)
legend1 = axes[0].legend(*scatter1.legend_elements(), title="True Categories")
axes[0].add_artist(legend1)
axes[0].set_title('Représentation par catégories reélles', fontsize=24)  # Augmenter la taille du titre
axes[0].set_xlabel('t-SNE Component 1')
axes[0].set_ylabel('t-SNE Component 2')

# Sous-graphique 2 : Projection des clusters prédits
scatter2 = axes[1].scatter(X_tsne[:, 0], X_tsne[:, 1], c=y_pred, cmap='viridis', alpha=0.6)
legend2 = axes[1].legend(*scatter2.legend_elements(), title="Clusters")
axes[1].add_artist(legend2)
axes[1].set_title('Représentation par clusters', fontsize=24)  # Augmenter la taille du titre
axes[1].set_xlabel('t-SNE Component 1')
axes[1].set_ylabel('t-SNE Component 2')

plt.tight_layout()
plt.show()
```

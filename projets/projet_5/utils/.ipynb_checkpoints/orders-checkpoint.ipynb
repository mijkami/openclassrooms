### get_wait_time
Return a dataframe with [order_id, wait_time, expected_wait_time, delay_vs_expected, order_status]

Hints:

```python
# Inspect orders dataframe
orders
```

```python
# Filter dataframe on delivered orders
orders.sort_values('order_delivered_customer_date')
```

```python
# handle datetime
orders['order_delivered_customer_date'] = pd.to_datetime(orders['order_delivered_customer_date'])
orders['order_purchase_timestamp'] = pd.to_datetime(orders['order_purchase_timestamp'])
orders['order_estimated_delivery_date'] = pd.to_datetime(orders['order_estimated_delivery_date'])
orders
```

```python
# compute wait time
# wait_time (float) the number of days between order_date and delivered_date
orders['wait_time'] = (orders['order_delivered_customer_date'] - orders['order_purchase_timestamp']).dt.days
orders
```

```python
orders.dtypes
```

```python
# compute expected wait time
# expected_wait_time (float) the number of days between order_date and estimated_delivery_date
orders['expected_wait_time'] = (orders['order_estimated_delivery_date'] - orders['order_purchase_timestamp']).dt.days
orders
```

```python
orders['delay_vs_expected'] = 0
orders
```

```python
orders[('order_id')][0]
```

```python
# delay_vs_expected (float): if the actual delivery date is later 
# than the estimated delivery date, returns the absolute number 
# of days between the two dates, otherwise return 0
orders['delay_vs_expected']  = orders['wait_time'] - orders['expected_wait_time']

def positive_value(d):
    return d if d > 0 else 0

orders['delay_vs_expected'] = orders['delay_vs_expected'].map(positive_value)


#for index, row in orders.iterrows():
#   if orders['order_estimated_delivery_date'][index] > orders['order_delivered_customer_date'][index]:
#        orders['delay_vs_expected'][index] = abs(orders['expected_wait_time'][index] - orders['wait_time'][index])
```

### get_review_score
     Returns a DataFrame with:
        order_id, dim_is_five_star, dim_is_one_star, review_score

```python
# dim_is_five_star (int) 1 if the order received a five_star, 0 otherwise
# dim_is_one_star (int) 1 if the order received a one_star, 0 otherwise
# review_score(int) from 1 to 5

reviews = data['order_reviews'].copy()
reviews
```

```python
# Fill in the functions below, which you will have to apply "element-wise" to each Series in the next cell below
# So as to create the 2 new columns requested 

# dim_is_five_star (int) 1 if the order received a five_star, 0 otherwise
def dim_five_star(d):
    return 1 if d == 5 else 0


# dim_is_one_star (int) 1 if the order received a one_star, 0 otherwise
def dim_one_star(d):
    return 1 if d == 1 else 0
```

```python
reviews["dim_is_five_star"] = reviews["review_score"].map(dim_five_star) # --> Series([0, 1, 1, 0, 0, 1 ...])


reviews["dim_is_one_star"] = reviews["review_score"].map(dim_one_star) # --> Series([0, 1, 1, 0, 0, 1 ...])
```

```python
# Check your new dataframe and commit your code to olist/order.py when it works. 
# Test it below
from olist.order import Order
Order().get_review_score()
```

```python
reviews = reviews[['order_id', 'dim_is_five_star', 'dim_is_one_star', 'review_score']]
reviews
```
